```python
## Build a List of Earthquake Objects
This section demonstrates lists, loops, and object creation from the dataset.
```

```python
earthquake_objects = []

for _, row in df.iterrows():
    event = EarthquakeEvent(
        magnitude=row["mag"],
        depth=row["depth"],
        location=row["place"],
        timestamp=row["time"],
        latitude=row["latitude"],
        longitude=row["longitude"]
    )
    earthquake_objects.append(event)

print("Number of EarthquakeEvent objects created:", len(earthquake_objects))
print(earthquake_objects[0])
```

```python
## Severity Classification Using Loops and If Statements
We classify each event and count how many earthquakes fall into each severity group.
```

```python
severity_counts = {"Minor": 0, "Moderate": 0, "Severe": 0}

for event in earthquake_objects:
    severity = event.classify_severity()
    if severity in severity_counts:
        severity_counts[severity] += 1

severity_counts
```

```python
## Standalone Function: Risk Score
The `compute_risk_score()` function calculates a normalized earthquake risk score using magnitude, depth, and population density.
```

```python
risk_scores = []

for event in earthquake_objects[:10]:
    score = compute_risk_score(
        magnitude=event.magnitude,
        depth=event.depth,
        population_density=1000
    )
    risk_scores.append(score)

risk_scores
```
